```python
import pandas as pd
data ={
    "Name": ["Naim","Rakib","Raju","Sakib","Joy","Roy","Ali","Jahan"],
    'City':['Dhaka','Rajshahi','Gaibandha','Los Angeles','Newark','Pabna','Nilphamari',"Old Dhaka"],
    'Department':["HR","IT",'HR','MANAGER',"Marketing","Worker","IT","selles"],
    'Salary':[10000,8000,3000,7000,11000,5000,9000,11500],
}


df= pd.DataFrame(data)
df
```

```python
df.loc[df['City'].str.contains("New")]
```

Case sensitive :

```python
df.loc[df['City'].str.contains("new")]    #Wron answer Because Case sensitive 
```

Solve : 

```python
df.loc[df['City'].str.contains("new",case=False)] 
```

COntain a word :

```python
#Exist a word : 

df.loc[df['City'].str.contains(r"Dha")]
```

Only Start with a Word :

```python

df.loc[df['City'].str.contains(r"^Dha")]
```

 End  with :

```python
#End with rk :
df.loc[df['City'].str.contains(r"rk$")]
```

start with a Vowel:

```python
df.loc[df['Name'].str.contains(r"^[aEIOU]",case=False)]
```

String Contain New OR Los :

```python
df.loc[df['City'].str.contains(r"New|Los")]
```
