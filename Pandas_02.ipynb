List to Data_Frame : 

```python
import pandas as pd

my_list=[['Naim',22],['Raju',23],['sakib',24]]
list_df=pd.DataFrame(my_list)
print(type(list_df))
list_df
```

columns_Adding : ['Name','Mid_marks']

```python
import pandas as pd

my_list=[['Naim',22],['Raju',23],['sakib',24]]
list_df=pd.DataFrame(my_list,columns=['Name','Mid_marks'])
print(type(list_df))
list_df
```

Index_Adding : [1,2,3]

```python
import pandas as pd

my_list=[['Naim',22],['Raju',23],['sakib',24]]
list_df=pd.DataFrame(my_list,columns=['Name','Mid_marks'],index=[1,2,3])
print(type(list_df))
list_df
```

Touple to Dataframe :

```python
import pandas as pd

my_tuple=(('Rakib',23),('Rajib',22),('sakhu',21))

df_tuple=pd.DataFrame(my_tuple,columns=['Name','Mid_marks'],index=[1,2,3])

print(type(df_tuple))
df_tuple
```

```python
import pandas as pd
my_dic ={
    'Name':['rakib','rajib','joy'], #key=>columns
    'Age':[21,23,12] #must be same length
} 
dic_df=pd.DataFrame(my_dic)
print(type(dic_df)) 
dic_df
```

```python
type(df['FullName']) #column/row=series     Multiple serise made a data frame
```

```python
#multiple row (in a normal list)
df.loc[[2,3,12,15]]
```

```python
#multiple row range [3 to 7 print 

df.loc[3:7] #[3 to 7 print ]
```

```python
df.loc[:,'FullName']
```
