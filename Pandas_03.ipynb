Changing index and columns:(iloc)

```python
import pandas as pd 

df=pd.read_csv('student_data.csv')
df
```

```python
#chnage the index
df_index=df.set_index('StudentID') 
df_index
```

```python
#df_index.iloc[3:8(range : n-1)]      #it will gives error because now our index is student_id
df_index.iloc[3:8]   

# we need to use iloc
```

```python
df_index.iloc[:,1:3]
```

```python
df_index.iloc[:,:]
```

```python
df.rename(columns={'FullName':'Full Name'},inplace=True)
df
```

Delet row , columns and changing values  :

```python
#delet row , columns and changing values 
#row delet 0,2,19
df.drop(0,inplace=True) 
df
```

```python
#column delet
df.drop('Python Marks',axis=1,inplace=True)
df
```

Value asign :

```python
df.loc[1,'Data Structure Marks']=90

df.head(5)
```

```python
df.loc[1,'CompletionStatus']='Completed'

df.head(5)
```

Adding to in a Range :

```python
df.loc[1:4,'Data Structure Marks']+=2
df.head(5)
```

```python
for i , series in df.iterrows(): 
    print(f"{i} : {series}")
```

```python
#show like touples 
for i in df.itertuples():
    print(i)
```

```python
for i in df.itertuples(index=False):
    print(i)
```
