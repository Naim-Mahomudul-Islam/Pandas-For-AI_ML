```python
import numpy as np

df.loc[df['FullName'] == 'Afsana Mimi', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Sadia Chowdhury', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Alif Rahman', 'Total Marks'] = np.nan

df
```

instractor wise divide student and then strat operation :

```python
Instructor_group =df.groupby('Instructor')
Instructor_group #key with data frame
```

```python
#for key value payer we need loop

for instructor , df_ins in  Instructor_group :
    print(instructor)
    print(df_ins)
```

```python
#for key value payer we need loop

for instructor , df_ins in  Instructor_group :
    print(instructor)
    display(df_ins)
```

```python
mean_marks = Instructor_group['Total Marks'].mean()
mean_marks
```

```python
#At first go to Find the Null Value
# the go to the iinstraction and find the mean value for the null
# and the initialize it 

df['Total Marks'] = df['Total Marks'].fillna(df['Instructor'].map(mean_marks))

df
```
